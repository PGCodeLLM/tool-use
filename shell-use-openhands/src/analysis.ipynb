```python
import re

def extract_str_replace_tool_interactions(messages: list[dict], 
                                          cmds_to_include: list[str] = None) -> list[tuple[str,str]]:
    """
    Extract successful `str_replace_editor` tool interactions;
    You can specify which type of commands you want with `cmds_to_include` list, default to all
    """

    cmd_interactions = []
    if not messages:
        return None

    # trajectory must be bigger than 3 interactions
    if len(messages) < 4:
        return None
    
    for idx in range(1, len(messages) - 1):
        msg = messages[idx]

        # skip non-assistant messages or ones w/o str_replace_editor tool calls
        if msg['role'] != 'assistant' or '<function=str_replace_editor>' not in msg['content']:
            continue
        
        # get the tool call context message (previous) and tool call result msg (next one)
        prev_msg_txt = messages[idx-1]['content']
        next_msg_txt = messages[idx+1]['content']

        # skip if this tool call failed
        if 'ERROR:' in next_msg_txt:
            continue

        # save context message and tool call message tuple
        cmd_interactions.append((prev_msg_txt, msg['content']))

        # TODO: filter based on the cmds_to_include

        # extract function call
        # match = re.search(r"(<function=str_replace_editor>.*?</function>)", msg['content'], re.DOTALL)
        # if match:
        #     func_call = match.group(1).strip()
        # else: 
        #     # no valid str_replace_editor call 
        #     continue
    
    return cmd_interactions
```
